## Heightfield dimensions Results


```python
import pandas as pd

dimension_data = pd.read_csv('./timeResults2D/results.csv')
dimension_data
```

```python
import altair as alt

```

```python
base = alt.Chart(dimension_data).mark_line(point=alt.OverlayMarkDef(filled=False, fill="white")).encode(
    x=alt.X('Particles:Q', title='Particles'),
    y=alt.Y('Time:Q', title='Time (ms)'),
    tooltip= ['Particles:Q', alt.Tooltip('Time:Q', title='ms')],
).properties(
    title='Passport strength by Year'
).transform_filter(
    alt.FieldEqualPredicate(field='predicted', equal=False)
)
```
