# Decomposition Experiments

## Heightfield dimensions


```python
import pandas as pd

data = pd.read_csv('./timeResults2D/results.csv')
data['Time per Particle'] = data['Time']/data['Particles'] 
data
```

```python
import altair as alt
base = alt.Chart(data).mark_line(point=alt.OverlayMarkDef(filled=False, fill="white")).encode(
    x=alt.X('Particles:Q', title=''),
    y=alt.Y('Time:Q', title='Time (ms)'),
    tooltip= ['Particles:N', alt.Tooltip('Time:Q', title='Time (ms)')],
).properties(
    title='Particles vs Time'
)
base
```

```python

base = alt.Chart(data).mark_line(point=alt.OverlayMarkDef(filled=False, fill="white")).encode(
    x=alt.X('Heightfield Dimensions:Q', title=''),
    y=alt.Y('Particles:Q', title='Particles'),
    tooltip= ['Particles:N', alt.Tooltip('Time:Q', title='Time (ms)')],
).properties(
    title='Dimensions vs Particles'
)
base
```

```python



base = alt.Chart(data).mark_line(point=alt.OverlayMarkDef(filled=False, fill="white")).encode(
    x=alt.X('Heightfield Dimensions:Q', title=''),
    y=alt.Y('Time:Q', title='Time (ms)'),
    tooltip= ['Particles:N', alt.Tooltip('Time:Q', title='Time (ms)')],
).properties(
    title='Particles vs Time'
)
base
```

```python



base = alt.Chart(data).mark_line(point=alt.OverlayMarkDef(filled=False, fill="white")).encode(
    x=alt.X('Heightfield Dimensions:Q', title=''),
    y=alt.Y('Time per Particle:Q', title='ms/particle'),
    tooltip= ['Particles:N', alt.Tooltip('Time:Q', title='Time (ms)')],
).properties(
    title='Dimensions vs Time per Particle'
)
base
```
